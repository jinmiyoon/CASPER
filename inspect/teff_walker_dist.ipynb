```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def temp_dis(teff, coeff, exponent):
    #temp=teff+teff*(coeff*np.exp(-1.0*exponent)*np.random.rand(100))
    temp=teff+teff*(float(coeff)*10**(-float(exponent))*np.random.rand(100))
    
    return temp

def temp_dis_hist(temp):
    counts, bins = np.histogram(temp, 50)
    fig,ax=plt.subplots(1,2, figsize=(8,4))
    hist_plot= ax[0].hist(bins[:-1], bins, weights=counts, histtype='step')
    scatter_plot=ax[1].scatter(np.arange(0,100,1), temp)
    return 

#initial + initial * (2e-3*np.random.rand(100, len(initial)))
```
