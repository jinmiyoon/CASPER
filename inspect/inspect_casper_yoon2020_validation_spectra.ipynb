```python
fits_directory1 = "../inputs/spectra/test_spectra/"
fits_directory2 = "../inputs/spectra/bf-validation/"

FitsName=np.array([])
StarName =np.array([])
RA=np.array([])
DEC=np.array([])


for fileName in glob.glob(fits_directory1+'*.fits'):
    hdulist = fits.open(fileName)  
    #hdulist.info()
    #hdulist[0].header
    ''''''
    fitsName=fileName.replace(fits_directory1,'')
    #print(fitsName)
    FitsName = np.append(FitsName, fitsName)
    name = hdulist[0].header['OBJECT']
    print("fitsname = %s, starname= %s "%(fitsName, name))
    if name =='30314-067': name ='BPS CS30314-067'
    if name =='HE1310m0536': name ='HE1310-0536'
    StarName =np.append(StarName, name)
    if fitsName =='g77-61_blue_sum.fits' : 
        ra =hdulist[0].header['OBJRA']
        dec = hdulist[0].header['OBJDEC']
    else: 
        ra = hdulist[0].header['RA'] 
        dec = hdulist[0].header['DEC']
    #print("ra = %s, dec = %s" %(ra, dec) )
    ra
    dec
    equinox= hdulist[0].header['equinox']
    equinox
    if type(ra) is float : 
        coord = SkyCoord(ra, dec,frame='icrs', unit='deg')
    else: coord =SkyCoord(ra+' '+dec,unit=(u.hourangle, u.deg))
    print(coord)
    
    RA= np.append(RA,coord.to_string('decimal').split(' ')[0])
    DEC= np.append(DEC,coord.to_string('decimal').split(' ')[1])
    
    hdulist.close()
    
```
